```python

import numpy as np
#X = matriz [-1, 1; 0, 2; 2, 0; 3, 3]
X = np.array([[-1, 1], 
            [0, 2],
            [2, 0],
            [3, 3]])

#N = número de filas de X
N = X.shape[0]

#mean = media de los datos
mean = np.mean(X, axis=0)

#X_tilde = X - media
X_tilde = X - mean

#Sigma = matriz de covarianzas empírica calculada a partir de X_tilde
Sigma = (X_tilde.T @ X_tilde) / N


```
