**Entropía cruzada entre P y ^P, Y mismo espacio muestral**

**(Entropía cruzada de la predicción ^P relativa a la referencia)**


H(p, ^p) = -sum(p(y) log2 ^p(y)) = -E[log2 ^p(y)]

```python

hmu = np.linspace(1e-2, 1-1e-2, 100); mu = 0.31 * np.ones_like(hmu)
H = lambda mu, hmu: -mu*np.log2(hmu) -(1-mu)*np.log2(1-hmu)
print(f'H(0.31, 0.4) = {H(0.31, 0.4).round(4)}')
plt.figure(figsize=(2.5, 2.5)); plt.plot(hmu, H(mu, hmu)); plt.grid();
```
